###*chatGPT*, *copilot* y otras variantes de inteligencia artificial

Las inteligencias artificiales (IA), en particular los large language models (LLMs) pueden servir como un asistente personal que les permita aumentar la productividad.
Pueden ser muy útiles en esta materia para:
* Encontrar errores en un código.
* Dar ejemplos y explicar cómo se utilizan algunas librerías (por ejemplo para graficar).
* Agregar comentarios a un código y reformatearlo para hacerlo más claro o más sencillo.

Como cualquier información de otra fuente (Internet, Wikipedia, un artículo científico, un libro, un profesor), tiene alguna probabilidad de ser incorrecta. En el caso de los LLM actuales pueden dar una respuesta incorrecta de una manera muy convincente. Hay que usarlas con pesamiento crítico y verificando con otras fuentes.

Que no les pase como a [Steven Schwartz](https://www.clarin.com/internacional/unidos-abogado-debio-disculparse-presentar-antecedentes-falsos-creados-chatgpt-podria-recibir-dura-sancion_0_hf500FslVs.html)...

Les sugiero que escriban primero una versión del código antes de eventualmente usar la IA para resolver problemas. Si la IA escribe una parte del código, asegúrense de que lo entienden (le pueden pedir que lo explique...) y que está haciendo lo que necesitan que haga. Eso les va a permitir modificar el código, solucionar problemas y adaptarlo a situaciones diferentes.

Esperamos que aprovechen esta materia para mejorar sus capacidades de programación, pensamiento crítico, entendimiento de la física y conocimiento de las posibilidades que ofrece la física computacional.
La forma de lograrlo es enfrentándose a los problemas de las prácticas e interactuando con la cátedra y sus compañeros y compañeras.
